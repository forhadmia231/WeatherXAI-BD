```python
# ============================================================
# WeatherXAI-BD Optimized V3
# Humidity + Temperature + Wind Speed
# LSTM + Transformer + Hybrid LSTM-Transformer
# ============================================================

import os, time, zipfile, warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from scipy.stats import ttest_rel

import tensorflow as tf
from tensorflow.keras.models import Model, Sequential
from tensorflow.keras.layers import (
    Input, LSTM, Dense, Dropout, LayerNormalization,
    MultiHeadAttention, GlobalAveragePooling1D
)
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau


# =========================
# 1. Paths
# =========================

DATA_PATH = "/kaggle/input/datasets/mdforhadmia/weather-2021-2024-dataset/WEATHER_2021_2024_MERGED.csv"

OUT_DIR = "/kaggle/working/WeatherXAI_BD_Optimized_V3"
FIG_DIR = f"{OUT_DIR}/figures"
TABLE_DIR = f"{OUT_DIR}/tables"
MODEL_DIR = f"{OUT_DIR}/models"

for d in [OUT_DIR, FIG_DIR, TABLE_DIR, MODEL_DIR]:
    os.makedirs(d, exist_ok=True)


# =========================
# 2. Load Dataset
# =========================

df = pd.read_csv(DATA_PATH)

# Clean column names
df.columns = df.columns.str.upper().str.strip()

# Remove duplicate columns
df = df.loc[:, ~df.columns.duplicated()]

print("Dataset Columns:")
print(df.columns.tolist())


# ============================================================
# IMPORTANT:
# Your dataset contains:
# DAY, MONTH, YEAR, LAT, LON,
# HUMIDITY, TEMPERATURE, WIND_SPEED
# ============================================================

# Target variables
target_cols_original = [
    "HUMIDITY",
    "TEMPERATURE",
    "WIND_SPEED"
]

# Required columns
required_cols = [
    "DAY",
    "MONTH",
    "YEAR",
    "LAT",
    "LON",
    "HUMIDITY",
    "TEMPERATURE",
    "WIND_SPEED"
]

# Check missing columns
for col in required_cols:
    if col not in df.columns:
        raise ValueError(f"Missing column: {col}")


# =========================
# 3. Create DATE
# =========================

df["DATE"] = pd.to_datetime(
    df[["YEAR", "MONTH", "DAY"]]
)

# Sort by location and date
df = df.sort_values(
    ["LAT", "LON", "DATE"]
).reset_index(drop=True)


# =========================
# 4. Missing Value Handling
# =========================

df[target_cols_original] = (
    df.groupby(["LAT", "LON"])[target_cols_original]
      .transform(lambda x: x.interpolate())
)

df[target_cols_original] = (
    df.groupby(["LAT", "LON"])[target_cols_original]
      .transform(lambda x: x.ffill().bfill())
)

# Remove remaining missing values
df = df.dropna(
    subset=target_cols_original
).reset_index(drop=True)


# =========================
# 5. Cyclical Encoding
# =========================

# Month
df["MONTH_SIN"] = np.sin(
    2 * np.pi * df["MONTH"] / 12
)

df["MONTH_COS"] = np.cos(
    2 * np.pi * df["MONTH"] / 12
)

# Day
df["DAY_SIN"] = np.sin(
    2 * np.pi * df["DAY"] / 31
)

df["DAY_COS"] = np.cos(
    2 * np.pi * df["DAY"] / 31
)


# =========================
# 6. Lag + Rolling Features
# =========================

base_weather_cols = [
    "HUMIDITY",
    "TEMPERATURE",
    "WIND_SPEED"
]

lag_days = [1, 7]
rolling_windows = [7]


for col in base_weather_cols:

    # Lag features
    for lag in lag_days:

        df[f"{col}_LAG_{lag}"] = (
            df.groupby(["LAT", "LON"])[col]
              .shift(lag)
        )

    # Rolling features
    for window in rolling_windows:

        df[f"{col}_ROLL_MEAN_{window}"] = (
            df.groupby(["LAT", "LON"])[col]
              .rolling(window)
              .mean()
              .reset_index(level=[0, 1], drop=True)
        )

        df[f"{col}_ROLL_STD_{window}"] = (
            df.groupby(["LAT", "LON"])[col]
              .rolling(window)
              .std()
              .reset_index(level=[0, 1], drop=True)
        )


# Remove rows created by lag/rolling operations
df = df.dropna().reset_index(drop=True)


# =========================
# 7. Feature Columns
# =========================

feature_cols = (
    base_weather_cols
    +
    [
        "LAT",
        "LON",
        "MONTH_SIN",
        "MONTH_COS",
        "DAY_SIN",
        "DAY_COS"
    ]
    +
    [
        col for col in df.columns
        if "LAG" in col or "ROLL" in col
    ]
)

print("\nTotal Features:", len(feature_cols))
print("Feature Columns:")
print(feature_cols)


# =========================
# 8. Target Columns
# =========================

target_cols = [
    "HUMIDITY",
    "TEMPERATURE",
    "WIND_SPEED"
]

print("\nTarget Columns:")
print(target_cols)


# =========================
# 9. Save Statistics
# =========================

df[target_cols_original].describe().T.to_csv(
    f"{TABLE_DIR}/descriptive_statistics_original.csv"
)

df[target_cols_original].corr().to_csv(
    f"{TABLE_DIR}/correlation_matrix_original.csv"
)


# =========================
# 10. Correlation Heatmap
# =========================

corr = df[target_cols_original].corr()

plt.figure(figsize=(8, 6))

plt.imshow(
    corr,
    cmap="viridis"
)

plt.colorbar()

plt.xticks(
    range(len(target_cols_original)),
    target_cols_original,
    rotation=30
)

plt.yticks(
    range(len(target_cols_original)),
    target_cols_original
)

for i in range(len(target_cols_original)):
    for j in range(len(target_cols_original)):

        plt.text(
            j,
            i,
            f"{corr.iloc[i, j]:.3f}",
            ha="center",
            va="center",
            color="white"
        )

plt.title(
    "Correlation Heatmap of Weather Variables"
)

plt.tight_layout()

plt.savefig(
    f"{FIG_DIR}/correlation_heatmap.png",
    dpi=300
)

plt.show()


# =========================
# 11. Scaling
# =========================

df = df.sort_values(
    ["LAT", "LON", "DATE"]
).reset_index(drop=True)


# 80% chronological split
split_row = int(len(df) * 0.8)

train_df = df.iloc[:split_row].copy()
test_df = df.iloc[split_row:].copy()


feature_scaler = MinMaxScaler()
target_scaler = MinMaxScaler()


# Fit scalers ONLY on training data
feature_scaler.fit(
    train_df[feature_cols]
)

target_scaler.fit(
    train_df[target_cols]
)


# Scale complete dataset
df_scaled = df.copy()

df_scaled[feature_cols] = (
    feature_scaler.transform(
        df[feature_cols]
    )
)

df_scaled[target_cols] = (
    target_scaler.transform(
        df[target_cols]
    )
)


# =========================
# 12. Location-wise Sequence
# =========================

SEQ_LEN = 30


def create_location_sequences(
    df_scaled,
    feature_cols,
    target_cols,
    seq_len
):

    X = []
    y = []

    for (lat, lon), group in df_scaled.groupby(
        ["LAT", "LON"]
    ):

        group = group.sort_values("DATE")

        features = group[
            feature_cols
        ].values

        targets = group[
            target_cols
        ].values

        if len(group) <= seq_len:
            continue

        for i in range(
            len(group) - seq_len
        ):

            X.append(
                features[
                    i:i + seq_len
                ]
            )

            y.append(
                targets[
                    i + seq_len
                ]
            )

    return np.array(X), np.array(y)


X, y = create_location_sequences(
    df_scaled,
    feature_cols,
    target_cols,
    SEQ_LEN
)


# =========================
# 13. Train-Test Split
# =========================

split = int(len(X) * 0.7)

X_train = X[:split]
X_test = X[split:]

y_train = y[:split]
y_test = y[split:]


print("\nSequence Shapes:")
print("X:", X.shape)
print("y:", y.shape)

print("\nTraining:")
print("X_train:", X_train.shape)
print("y_train:", y_train.shape)

print("\nTesting:")
print("X_test:", X_test.shape)
print("y_test:", y_test.shape)


pd.DataFrame({
    "Dataset": [
        "X_train",
        "X_test",
        "y_train",
        "y_test"
    ],

    "Shape": [
        str(X_train.shape),
        str(X_test.shape),
        str(y_train.shape),
        str(y_test.shape)
    ]

}).to_csv(
    f"{TABLE_DIR}/train_test_dimensions.csv",
    index=False
)


# =========================
# 14. LSTM Model
# =========================

def build_lstm_model(
    seq_len,
    n_features,
    n_outputs
):

    model = Sequential([

        LSTM(
            96,
            return_sequences=True,
            input_shape=(
                seq_len,
                n_features
            )
        ),

        Dropout(0.25),

        LSTM(48),

        Dropout(0.25),

        Dense(
            64,
            activation="relu"
        ),

        Dense(n_outputs)

    ])

    model.compile(
        optimizer=Adam(0.0005),
        loss="mse"
    )

    return model


# =========================
# 15. Transformer Encoder
# =========================

def transformer_encoder(
    inputs,
    head_size=32,
    num_heads=2,
    ff_dim=128,
    dropout=0.2
):

    attn = MultiHeadAttention(
        num_heads=num_heads,
        key_dim=head_size
    )(inputs, inputs)

    attn = Dropout(dropout)(attn)

    x = LayerNormalization(
        epsilon=1e-6
    )(inputs + attn)

    ff = Dense(
        ff_dim,
        activation="relu"
    )(x)

    ff = Dropout(dropout)(ff)

    ff = Dense(
        inputs.shape[-1]
    )(ff)

    x = LayerNormalization(
        epsilon=1e-6
    )(x + ff)

    return x


# =========================
# 16. Transformer Model
# =========================

def build_transformer_model(
    seq_len,
    n_features,
    n_outputs
):

    inputs = Input(
        shape=(
            seq_len,
            n_features
        )
    )

    x = transformer_encoder(
        inputs,
        head_size=32,
        num_heads=2,
        ff_dim=128
    )

    x = GlobalAveragePooling1D()(x)

    x = Dense(
        64,
        activation="relu"
    )(x)

    x = Dropout(0.25)(x)

    outputs = Dense(n_outputs)(x)

    model = Model(
        inputs,
        outputs
    )

    model.compile(
        optimizer=Adam(0.0005),
        loss="mse"
    )

    return model


# =========================
# 17. Hybrid LSTM-Transformer
# =========================

def build_simple_hybrid_model(
    seq_len,
    n_features,
    n_outputs
):

    inputs = Input(
        shape=(
            seq_len,
            n_features
        )
    )

    x = LSTM(
        64,
        return_sequences=True
    )(inputs)

    x = Dropout(0.25)(x)

    x = transformer_encoder(
        x,
        head_size=32,
        num_heads=2,
        ff_dim=128,
        dropout=0.2
    )

    x = GlobalAveragePooling1D()(x)

    x = Dense(
        64,
        activation="relu"
    )(x)

    x = Dropout(0.25)(x)

    outputs = Dense(n_outputs)(x)

    model = Model(
        inputs,
        outputs
    )

    model.compile(
        optimizer=Adam(0.0005),
        loss="mse"
    )

    return model


# =========================
# 18. Create Models
# =========================

n_features = X_train.shape[2]
n_outputs = y_train.shape[1]

print("\nNumber of Features:", n_features)
print("Number of Outputs:", n_outputs)


models = {

    "LSTM":
        build_lstm_model(
            SEQ_LEN,
            n_features,
            n_outputs
        ),

    "Transformer":
        build_transformer_model(
            SEQ_LEN,
            n_features,
            n_outputs
        ),

    "Hybrid_LSTM_Transformer":
        build_simple_hybrid_model(
            SEQ_LEN,
            n_features,
            n_outputs
        )
}


# =========================
# 19. Training
# =========================

EPOCHS = 100
BATCH_SIZE = 32


early_stop = EarlyStopping(
    monitor="val_loss",
    patience=15,
    restore_best_weights=True
)


reduce_lr = ReduceLROnPlateau(
    monitor="val_loss",
    factor=0.5,
    patience=5,
    min_lr=1e-6
)


histories = []
training_times = []


for name, model in models.items():

    print(
        f"\nTraining {name}..."
    )

    start = time.time()

    history = model.fit(

        X_train,
        y_train,

        validation_split=0.2,

        epochs=EPOCHS,

        batch_size=BATCH_SIZE,

        callbacks=[
            early_stop,
            reduce_lr
        ],

        verbose=1
    )

    end = time.time()

    histories.append(history)

    training_times.append({

        "Model": name,

        "Training_Time_Seconds":
            end - start

    })

    model.save(
        f"{MODEL_DIR}/{name}.h5"
    )


pd.DataFrame(
    training_times
).to_csv(
    f"{TABLE_DIR}/training_time.csv",
    index=False
)


# =========================
# 20. Loss Curves
# =========================

for name, history in zip(
    models.keys(),
    histories
):

    plt.figure(
        figsize=(8, 5)
    )

    plt.plot(
        history.history["loss"],
        label="Training Loss"
    )

    plt.plot(
        history.history["val_loss"],
        label="Validation Loss"
    )

    plt.title(
        f"{name} Training and Validation Loss"
    )

    plt.xlabel("Epoch")
    plt.ylabel("Loss")

    plt.legend()

    plt.tight_layout()

    plt.savefig(
        f"{FIG_DIR}/{name}_loss_curve.png",
        dpi=300
    )

    plt.show()


# =========================
# 21. Inverse Transform
# =========================

def inverse_targets(y_scaled):

    inv = target_scaler.inverse_transform(
        y_scaled
    )

    return inv


# =========================
# 22. Metrics
# =========================

final_target_cols = [
    "HUMIDITY",
    "TEMPERATURE",
    "WIND_SPEED"
]


# Actual values
y_test_inv = inverse_targets(
    y_test
)


predictions = {}
results = []


for name, model in models.items():

    print(
        f"\nPredicting with {name}..."
    )

    pred_scaled = model.predict(
        X_test
    )

    pred_inv = inverse_targets(
        pred_scaled
    )

    predictions[name] = pred_inv


    # Metrics for each variable
    for i, var in enumerate(
        final_target_cols
    ):

        mae = mean_absolute_error(
            y_test_inv[:, i],
            pred_inv[:, i]
        )

        mse = mean_squared_error(
            y_test_inv[:, i],
            pred_inv[:, i]
        )

        rmse = np.sqrt(mse)

        r2 = r2_score(
            y_test_inv[:, i],
            pred_inv[:, i]
        )


        results.append({

            "Model": name,

            "Variable": var,

            "MAE": mae,

            "MSE": mse,

            "RMSE": rmse,

            "R2": r2

        })


metrics_df = pd.DataFrame(
    results
)


metrics_df.to_csv(
    f"{TABLE_DIR}/model_metrics.csv",
    index=False
)


print("\n========================")
print("MODEL METRICS")
print("========================")

print(metrics_df)


# =========================
# 23. Average R2
# =========================

avg_r2 = (
    metrics_df
    .groupby("Model")["R2"]
    .mean()
    .sort_values(
        ascending=False
    )
)


print("\nAverage R2:")
print(avg_r2)


best_model_name = avg_r2.index[0]

best_pred = predictions[
    best_model_name
]


print(
    f"\nBest Model: {best_model_name}"
)


# =========================
# 24. Model Comparison Graphs
# =========================

for metric in [
    "MAE",
    "RMSE",
    "R2"
]:

    pivot = metrics_df.pivot(
        index="Variable",
        columns="Model",
        values=metric
    )

    plt.figure(
        figsize=(10, 6)
    )

    pivot.plot(
        kind="bar",
        figsize=(10, 6)
    )

    plt.title(
        f"Model Comparison Based on {metric}"
    )

    plt.xlabel("Variable")

    plt.ylabel(metric)

    plt.xticks(
        rotation=20
    )

    plt.tight_layout()

    plt.savefig(
        f"{FIG_DIR}/model_comparison_{metric}.png",
        dpi=300
    )

    plt.show()


# =========================
# 25. Actual vs Predicted
# =========================

sample_n = min(
    500,
    len(y_test_inv)
)


for name, pred_inv in predictions.items():

    for i, var in enumerate(
        final_target_cols
    ):

        plt.figure(
            figsize=(12, 5)
        )

        plt.plot(
            y_test_inv[
                :sample_n,
                i
            ],
            label="Actual"
        )

        plt.plot(
            pred_inv[
                :sample_n,
                i
            ],
            label="Predicted"
        )

        plt.title(
            f"{name}: Actual vs Predicted {var}"
        )

        plt.xlabel(
            "Test Sample"
        )

        plt.ylabel(var)

        plt.legend()

        plt.tight_layout()

        plt.savefig(
            f"{FIG_DIR}/{name}_actual_vs_predicted_{var}.png",
            dpi=300
        )

        plt.show()


# =========================
# 26. Statistical Test
# =========================

hybrid_pred = predictions[
    "Hybrid_LSTM_Transformer"
]


stat_rows = []


for model_name in [
    "LSTM",
    "Transformer"
]:

    pred = predictions[
        model_name
    ]

    for i, var in enumerate(
        final_target_cols
    ):

        err_model = np.abs(
            y_test_inv[:, i]
            - pred[:, i]
        )

        err_hybrid = np.abs(
            y_test_inv[:, i]
            - hybrid_pred[:, i]
        )

        stat, p_value = ttest_rel(
            err_model,
            err_hybrid
        )


        stat_rows.append({

            "Comparison":
                f"{model_name} vs Hybrid",

            "Variable":
                var,

            "t_statistic":
                stat,

            "p_value":
                p_value,

            "Significant":
                "Yes"
                if p_value < 0.05
                else "No"

        })


stat_df = pd.DataFrame(
    stat_rows
)


stat_df.to_csv(
    f"{TABLE_DIR}/statistical_test.csv",
    index=False
)


print("\n========================")
print("STATISTICAL TEST")
print("========================")

print(stat_df)


# =========================
# 27. Agriculture Risk
# =========================

risk_df = pd.DataFrame(
    best_pred,
    columns=[
        f"PRED_{c}"
        for c in final_target_cols
    ]
)


# Heatwave
risk_df["Heatwave"] = (
    risk_df["PRED_TEMPERATURE"] > 35
)


# High Wind
risk_df["High_Wind"] = (
    risk_df["PRED_WIND_SPEED"] > 10
)


# Drought risk
# Since precipitation is NOT available,
# drought risk cannot use precipitation.
#
# Here we use:
# Low humidity + high temperature

risk_df["Drought_Risk"] = (
    (risk_df["PRED_HUMIDITY"] < 15)
    &
    (risk_df["PRED_TEMPERATURE"] > 30)
)


# Agriculture Risk Score
def agriculture_risk(row):

    score = (
        int(row["Heatwave"])
        +
        int(row["High_Wind"])
        +
        int(row["Drought_Risk"])
    )

    if score == 0:
        return "Low"

    elif score == 1:
        return "Medium"

    else:
        return "High"


risk_df["Agriculture_Risk"] = (
    risk_df.apply(
        agriculture_risk,
        axis=1
    )
)


risk_df.to_csv(
    f"{TABLE_DIR}/agriculture_risk.csv",
    index=False
)


# =========================
# 28. Risk Distribution
# =========================

plt.figure(
    figsize=(7, 5)
)

risk_df[
    "Agriculture_Risk"
].value_counts().plot(
    kind="bar"
)

plt.title(
    "Agricultural Risk Level Distribution"
)

plt.xlabel(
    "Risk Level"
)

plt.ylabel(
    "Count"
)

plt.xticks(
    rotation=0
)

plt.tight_layout()

plt.savefig(
    f"{FIG_DIR}/agriculture_risk_distribution.png",
    dpi=300
)

plt.show()


# =========================
# 29. Extreme Event Count
# =========================

event_counts = {

    "Heatwave":
        risk_df["Heatwave"].sum(),

    "High Wind":
        risk_df["High_Wind"].sum(),

    "Drought Risk":
        risk_df["Drought_Risk"].sum()

}


plt.figure(
    figsize=(8, 5)
)

plt.bar(
    event_counts.keys(),
    event_counts.values()
)

plt.title(
    "Extreme Weather Event Detection Count"
)

plt.xlabel(
    "Event"
)

plt.ylabel(
    "Count"
)

plt.xticks(
    rotation=20
)

plt.tight_layout()

plt.savefig(
    f"{FIG_DIR}/extreme_event_count.png",
    dpi=300
)

plt.show()


# =========================
# 30. ZIP Results
# =========================

zip_path = (
    "/kaggle/working/"
    "WeatherXAI_BD_Optimized_V3.zip"
)


with zipfile.ZipFile(
    zip_path,
    "w",
    zipfile.ZIP_DEFLATED
) as zipf:

    for root, dirs, files in os.walk(
        OUT_DIR
    ):

        for file in files:

            file_path = os.path.join(
                root,
                file
            )

            arcname = os.path.relpath(
                file_path,
                OUT_DIR
            )

            zipf.write(
                file_path,
                arcname
            )


print("\n========================")
print("DONE!")
print("========================")

print(
    "ZIP File:",
    zip_path
)
```
